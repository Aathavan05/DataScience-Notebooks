```python
#Data Visualization Implementation with a Unique Dataset
#Dataset: Smart City Traffic and Air Quality Analysis

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D


#create a Unique Dataset

np.random.seed(42)

n = 100

data = pd.DataFrame({
    "Day": np.arange(1, n + 1),
    "Traffic_Volume": np.random.randint(200, 1000, n),
    "Average_Speed": np.random.randint(20, 80, n),
    "Air_Quality_Index": np.random.randint(50, 200, n),
    "Temperature": np.random.uniform(25, 40, n)
})

print(data.head())

#Step 2: Scatter Plot
plt.figure()
plt.scatter(data["Traffic_Volume"], data["Air_Quality_Index"])
plt.title("Scatter Plot: Traffic Volume vs Air Quality Index")
plt.xlabel("Traffic Volume")
plt.ylabel("Air Quality Index")
plt.show()

#Step 3: Line Plot
plt.figure()
plt.plot(data["Day"], data["Traffic_Volume"])
plt.title("Line Plot: Traffic Volume Over Days")
plt.xlabel("Day")
plt.ylabel("Traffic Volume")
plt.show()

#Step 4: Histogram
plt.figure()
plt.hist(data["Average_Speed"], bins=10)
plt.title("Histogram: Distribution of Average Speed")
plt.xlabel("Average Speed")
plt.ylabel("Frequency")
plt.show()

#Step 5: Box Plot
plt.figure()
sns.boxplot(y=data["Air_Quality_Index"])
plt.title("Box Plot: Air Quality Index")
plt.show()

#Step 6: Pair Plot
sns.pairplot(data)
plt.show()

#Step 7: 3D Scatter Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    data["Traffic_Volume"],
    data["Average_Speed"],
    data["Air_Quality_Index"]
)

ax.set_title("3D Scatter Plot")
ax.set_xlabel("Traffic Volume")
ax.set_ylabel("Average Speed")
ax.set_zlabel("Air Quality Index")

plt.show()

#Step 8: 3D Line Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(
    data["Day"],
    data["Traffic_Volume"],
    data["Temperature"]
)

ax.set_title("3D Line Plot")
ax.set_xlabel("Day")
ax.set_ylabel("Traffic Volume")
ax.set_zlabel("Temperature")

plt.show()

#Step 9: 3D Surface Plot
X = np.linspace(1, 10, 30)
Y = np.linspace(1, 10, 30)
X, Y = np.meshgrid(X, Y)
Z = np.sin(X) + np.cos(Y)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_title("3D Surface Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Value")

plt.show()
```
